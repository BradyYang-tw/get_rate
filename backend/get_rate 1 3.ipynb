```python
import requests 
import numpy as np 
import pandas as pd 
from bs4 import BeautifulSoup

# 取得基金公司的code
url = 'https://www.moneydj.com/funddj/ya/YP081000List.djhtm?a=2'
r = requests.get(url)
soup = BeautifulSoup(r.text, "lxml")
td_tags = soup.find_all("td")
a_tags = [td.find('a') for td in td_tags if td.find('a') and td.find('a').get('href').startswith('/funddj/ya/yp081001.djhtm?')]
companys = []
for code in a_tags:
    if code.get('href').split('?')[-1] not in companys:
        companys.append(code.get('href').split('?')[-1])

# 取得基金code
fund_name = []
fund_code = []
for company in companys:
    url = f'https://www.moneydj.com/funddj/ya/yp081001.djhtm?{company}&ff=1'
    r = requests.get(url)
    soup = BeautifulSoup(r.text, "lxml")
    for fund in soup.find_all("td", {"class": "t3t1c1"}):
        fund_name .append(fund.text)
        fund_code.append(fund.find('a').get('href').split('?')[-1])

# 取得基金規模
fund_size = []
fund_dividend = []
sharpe = []
beta = []
fail_code = []
count = 0
for fund in fund_code:
    try:
        url = f'https://www.moneydj.com/funddj/yp/yp011001.djhtm?{fund}'
        r = requests.get(url)
        soup = BeautifulSoup(r.text, "lxml")
        # print(soup.find_all("td", {"class": "t3t2"})[3])
        fund_size.append(soup.find_all("td", {"class": "t3t2"})[3])
    except:
        print(f'{fund} 取得基金規模失敗', url)
    try:
        # 取得年化配息率
        url = f'https://www.moneydj.com/funddj/yp/wb05.djhtm?{fund}'
        data = pd.read_html(url)
        # print(data[3]['年化配息率(%)'][0])
        for i in data:
            if '年化配息率%' in i:   
                fund_dividend.append(i['年化配息率%'][0])
    except:
        print(f'{fund} 取得年化配息率失敗', url)
    try:
        # 取得Sharp Beta
        url = f'https://www.moneydj.com/funddj/yp/yp012001.djhtm?{fund}'
        data = pd.read_html(url)
        for i in data:
            if 'Sharpe' in i.values:
                sharpe.append(i[5][1])
                beta.append(i[6][1])
    except:
        print(f'{fund} 取得Sharp Beta失敗', url)
    count += 1
    # if count == 10:
    #     break
    
```

```python
fund_df = pd.DataFrame(list(zip(fund_name, fund_code, fund_dividend, sharpe, beta)), columns =['fund_name', 'fund_code', 'fund_dividend', 'sharpe', 'beta'])
```

```python
fund_df.to_csv('./fund.csv',index=False, encoding='utf-8-sig')
```
